# Data Overview

```python
# libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# csv reading function for all files in data
def data_reading(file_path):
    data = pd.read_csv("data/" + file_path)
    return data
```

```python
# read all files in data folder
diagnosis = data_reading("diagnosis.csv")
edstays = data_reading("edstays.csv")
medreconc = data_reading("medrecon.csv")
pyxis = data_reading("pyxis.csv")
triage = data_reading("triage.csv")
vitalsign = data_reading("vitalsign.csv")

patient_info = {
    "diagnosis": diagnosis,
    "edstays": edstays,
    "medreconc": medreconc,
    "pyxis": pyxis,
    "triage": triage,
    "vitalsign": vitalsign
}
```

```python
# iterate through df and print info and shape
for key, value in patient_info.items():
    print(f"{key} info:")
    print(value.info())
    print(f"{key} shape: {value.shape}")
    print("\n")
```

```python
# check if all stay_ids are unique in diagnosis
print("Are all stay_ids unique in diagnosis?", diagnosis['stay_id'].is_unique)

# print all duplicate stay_ids in diagnosis and the total number
duplicate_stay_ids = diagnosis[diagnosis.duplicated(subset=['stay_id'], keep=False)]['stay_id']
print("Duplicate stay_ids in diagnosis:", duplicate_stay_ids.unique())
print("Total number of duplicate stay_ids:", len(duplicate_stay_ids.unique()))
```

```python
# for every unique stay_id in duplicate_stays, check that there is only one unique subject_id
for stay_id in duplicate_stays['stay_id'].unique():
    subject_ids = duplicate_stays[duplicate_stays['stay_id'] == stay_id]['subject_id'].unique()
    if len(subject_ids) > 1:
        print(f"Stay ID {stay_id} has multiple subject IDs: {subject_ids}")
    else:
        print(f"Stay ID {stay_id} has one unique subject ID: {subject_ids[0]}")
```
