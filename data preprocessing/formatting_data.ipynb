```python
# reformat merged_data to have one row per stay_id, with UNIQUE values in the rest of the columns put in array
reformatted_data = merged_data.groupby('stay_id').agg(lambda x: list(set(x))).reset_index()
```

```python
# function that calculates difference between each element in charttime and intime
def calculate_time_diff(charttime_list, intime_list):
    time_diffs = []
    for charttime in charttime_list:
        for intime in intime_list:
            time_diff = pd.to_datetime(charttime) - pd.to_datetime(intime)
            time_diffs.append(time_diff.total_seconds() / 3600)  # convert to hours
            # round to three decimal places
            time_diffs[-1] = round(time_diffs[-1], 4)
    return time_diffs
```

```python
# create new columns charttime_intime_diff and charttime_vital_intime_diff in reformatted_data
reformatted_data['charttime_intime_diff'] = reformatted_data.apply(lambda row: calculate_time_diff(row['charttime'], row['intime']), axis=1)
reformatted_data['charttime_vital_intime_diff'] = reformatted_data.apply(lambda row: calculate_time_diff(row['charttime_vital'], row['intime']), axis=1)

# total time visit
reformatted_data['total_time_visit'] = reformatted_data.apply(lambda row: calculate_time_diff(row['outtime'], row['intime']), axis=1)

# drop charttime and charttime_vital columns
reformatted_data = reformatted_data.drop(columns=['charttime', 'charttime_vital', 'intime', 'outtime'])
```

```python
import ast
import numpy as np

# Work on baseline_input only; keep icd_title and all other non-listed columns unchanged.

# 1) Drop admmited_charttime
if 'admmited_charttime' in baseline_input.columns:
    baseline_input = baseline_input.drop(columns=['admmited_charttime'])


def _safe_literal_eval(value):
    """Safely parse string-encoded Python literals using ast.literal_eval."""
    try:
        if isinstance(value, str):
            value = value.strip()
            if value == '':
                return np.nan
            return ast.literal_eval(value)
        return value
    except (ValueError, SyntaxError, TypeError):
        return np.nan


def _to_float_list(value):
    """Parse into a list of floats; return np.nan when invalid."""
    parsed = _safe_literal_eval(value)

    if isinstance(parsed, (list, tuple, np.ndarray)):
        out = []
        for x in parsed:
            try:
                out.append(float(x))
            except (TypeError, ValueError):
                continue
        return out if len(out) > 0 else np.nan

    # If already numeric, keep as single-item list
    if isinstance(parsed, (int, float, np.integer, np.floating)):
        return [float(parsed)]

    return np.nan


def _mean_from_list(value):
    """Parse list-like value and flatten to mean float."""
    vals = _to_float_list(value)
    if isinstance(vals, list) and len(vals) > 0:
        return float(np.mean(vals))
    return np.nan


def _single_float(value):
    """Parse single-element list (or numeric) into one float."""
    vals = _to_float_list(value)
    if isinstance(vals, list) and len(vals) > 0:
        return float(vals[0])
    return np.nan


# 2) charttime_intime_diff: keep as parsed float list (array-like), no mean, no NA fill
if 'charttime_vital_intime_diff' in baseline_input.columns:
    baseline_input['charttime_vital_intime_diff'] = baseline_input['charttime_vital_intime_diff'].apply(_to_float_list)

# 3) charttime_vital_intime_diff: same cleaning
if 'charttime_intime_diff' in baseline_input.columns:
    baseline_input['charttime_intime_diff'] = (
        baseline_input['charttime_intime_diff']
        .apply(_mean_from_list)
        .astype(float)
    )
    baseline_input['charttime_intime_diff'] = baseline_input['charttime_intime_diff'].fillna(
        baseline_input['charttime_intime_diff'].median()
    )

# 4) temperature: single-element list string -> float, then fill NA with median
if 'temperature' in baseline_input.columns:
    baseline_input['temperature'] = baseline_input['temperature'].apply(_single_float).astype(float)
    baseline_input['temperature'] = baseline_input['temperature'].fillna(baseline_input['temperature'].median())

# 7) Vital sign columns: parse string-encoded lists into real float lists
vital_list_columns = [
    'heartrate',
    'resprate',
    'o2sat',
    'sbp',
    'dbp',
    'heartrate_vital',
    'resprate_vital',
    'o2sat_vital',
    'sbp_vital',
    'dbp_vital',
]

for col in vital_list_columns:
    if col in baseline_input.columns:
        baseline_input[col] = baseline_input[col].apply(_to_float_list)
```

```python
# Clean pain column in baseline_input
if 'pain' in baseline_input.columns:
    def _safe_parse_list(val):
        try:
            if isinstance(val, str):
                val = val.strip()
                if val == '':
                    return []
                parsed = ast.literal_eval(val)
            else:
                parsed = val
        except (ValueError, SyntaxError, TypeError):
            return []

        if isinstance(parsed, (list, tuple, np.ndarray)):
            return list(parsed)
        return [parsed]

    def _clean_pain_element(x):
        try:
            v = float(x)
        except (ValueError, TypeError):
            return np.nan

        if v < 0:
            return np.nan
        if v > 10:
            return 10.0
        return float(v)

    def _pain_mean_from_list(val):
        items = _safe_parse_list(val)
        cleaned = [_clean_pain_element(x) for x in items]

        valid = [x for x in cleaned if not pd.isna(x)]
        if len(valid) == 0:
            return np.nan
        return float(np.mean(valid))

    baseline_input['pain'] = baseline_input['pain'].apply(_pain_mean_from_list).astype(float)

    pain_median = baseline_input['pain'].median()
    baseline_input['pain'] = baseline_input['pain'].fillna(pain_median)

    # Ensure bounded numeric output and no missing values.
    baseline_input['pain'] = baseline_input['pain'].clip(lower=0.0, upper=10.0)
```

```python
import ast
import numpy as np

# Convert selected list-encoded columns to single float values (first element only)
cols_to_flatten = [
    'temperature',
    'heartrate',
    'resprate',
    'o2sat',
    'sbp',
    'dbp',
    'pain',
    'acuity',
    'icd_version',
    'total_time_visit'
]


def _parse_first_float(value):
    try:
        parsed = ast.literal_eval(value) if isinstance(value, str) else value
    except (ValueError, SyntaxError, TypeError):
        return np.nan

    # Handle list-like parsed values; take first element only.
    if isinstance(parsed, (list, tuple, np.ndarray)):
        if len(parsed) == 0:
            return np.nan
        first = parsed[0]
    else:
        first = parsed

    try:
        return float(first)
    except (ValueError, TypeError):
        return np.nan


for col in cols_to_flatten:
    if col in baseline_input.columns:
        baseline_input[col] = baseline_input[col].apply(_parse_first_float).astype(float)
        baseline_input[col] = baseline_input[col].fillna(baseline_input[col].median())
```

```python
# 1) Set stay_id as index
if 'stay_id' in baseline_input.columns:
    baseline_input = baseline_input.set_index('stay_id')

# 2) Drop icd_version
if 'icd_version' in baseline_input.columns:
    baseline_input = baseline_input.drop(columns=['icd_version'])

# 3) Create mean/trend features for selected vital/list columns
vital_cols = [
    'heartrate_vital',
    'resprate_vital',
    'o2sat_vital',
    'sbp_vital',
    'dbp_vital',
    'charttime_intime_diff',
    'charttime_vital_intime_diff',
]


def _parse_float_list(value):
    try:
        parsed = ast.literal_eval(value) if isinstance(value, str) else value
    except (ValueError, SyntaxError, TypeError):
        return np.nan

    if isinstance(parsed, (list, tuple, np.ndarray)):
        vals = []
        for x in parsed:
            try:
                vals.append(float(x))
            except (TypeError, ValueError):
                continue
        return vals if len(vals) > 0 else np.nan

    try:
        return [float(parsed)]
    except (TypeError, ValueError):
        return np.nan


def _list_mean(value):
    vals = _parse_float_list(value)
    if isinstance(vals, list) and len(vals) > 0:
        return float(np.mean(vals))
    return np.nan


def _list_trend(value):
    vals = _parse_float_list(value)
    if isinstance(vals, list) and len(vals) > 0:
        return float(vals[-1] - vals[0])
    return np.nan


for col in vital_cols:
    if col in baseline_input.columns:
        mean_col = f"{col}_mean"
        trend_col = f"{col}_trend"

        baseline_input[mean_col] = baseline_input[col].apply(_list_mean).astype(float)
        baseline_input[trend_col] = baseline_input[col].apply(_list_trend).astype(float)

        baseline_input[mean_col] = baseline_input[mean_col].fillna(baseline_input[mean_col].median())
        baseline_input[trend_col] = baseline_input[trend_col].fillna(baseline_input[trend_col].median())

        baseline_input = baseline_input.drop(columns=[col])
```

```python
from sklearn.preprocessing import LabelEncoder

# Encode selected categorical columns only
categorical_cols = ['gender', 'race', 'arrival_transport', 'disposition']
label_mappings = {}

for col in categorical_cols:
    if col in baseline_input.columns:
        le = LabelEncoder()

        # Treat missing values as their own category
        baseline_input[col] = baseline_input[col].fillna('Unknown').astype(str)

        # Encode in place
        baseline_input[col] = le.fit_transform(baseline_input[col])

        # Build interpretable mapping: label -> integer
        mapping = {label: int(idx) for idx, label in enumerate(le.classes_)}
        label_mappings[col] = mapping

# Print mappings for later interpretation
for col, mapping in label_mappings.items():
    print(f"{col}: {mapping}")
```

```python
import ast
import re
from collections import Counter

import numpy as np
import pandas as pd

# -------- Helpers --------
def _safe_parse_list(value):
    try:
        parsed = ast.literal_eval(value) if isinstance(value, str) else value
    except (ValueError, SyntaxError, TypeError):
        return []

    if isinstance(parsed, (list, tuple, np.ndarray)):
        return list(parsed)
    if pd.isna(parsed):
        return []
    return [parsed]


def _slugify(name):
    s = str(name).strip().lower()
    s = re.sub(r'[^a-z0-9]+', '_', s)
    s = re.sub(r'_+', '_', s).strip('_')
    return s


# -------- 1) med_record -> top 25 multi-hot --------
if 'med_record' in baseline_input.columns:
    med_lists = baseline_input['med_record'].apply(_safe_parse_list)

    med_counter = Counter()
    for meds in med_lists:
        cleaned = [str(m).strip() for m in meds if str(m).strip()]
        med_counter.update(cleaned)

    top25_med_record = [m for m, _ in med_counter.most_common(25)]

    for med in top25_med_record:
        col_name = f"medrec_{_slugify(med)}"
        baseline_input[col_name] = med_lists.apply(
            lambda meds: int(any(str(x).strip() == med for x in meds))
        ).astype(int)

    baseline_input = baseline_input.drop(columns=['med_record'])

    print('Top 25 med_record medications:')
    print(top25_med_record)


# -------- 2) adm_med_name -> normalized top 15 multi-hot --------
if 'adm_med_name' in baseline_input.columns:
    dose_pattern = re.compile(
        r'\d+(\.\d+)?\s*(mg|mcg|ml|g|%|units?|meq)[^\s]*',
        flags=re.IGNORECASE,
    )
    form_pattern = re.compile(
        r'\b(tab|vial|udcup|btl|soln|cap|inj|susp|syr)\b',
        flags=re.IGNORECASE,
    )

    def _normalize_adm_med(name):
        s = str(name).lower().strip()

        # Specific consolidation mappings BEFORE counting
        if 'morphine' in s:
            return 'morphine'
        if 'hydromorphone' in s:
            return 'hydromorphone'
        if 'ceftriaxone' in s:
            return 'ceftriaxone'
        if 'sodium chloride' in s:
            return 'sodium chloride'
        if 'pantoprazole' in s:
            return 'pantoprazole'
        if 'levothyroxine' in s:
            return 'levothyroxine'

        # Strip dosage-like tokens and form suffix words
        s = dose_pattern.sub(' ', s)
        s = form_pattern.sub(' ', s)

        # Cleanup punctuation/extra whitespace
        s = re.sub(r'[^a-z0-9\s]', ' ', s)
        s = re.sub(r'\s+', ' ', s).strip()
        return s

    adm_lists_raw = baseline_input['adm_med_name'].apply(_safe_parse_list)
    adm_lists_norm = adm_lists_raw.apply(
        lambda meds: [
            _normalize_adm_med(m)
            for m in meds
            if _normalize_adm_med(m) != ''
        ]
    )

    adm_counter = Counter()
    for meds in adm_lists_norm:
        adm_counter.update(meds)

    top15_adm_med = [m for m, _ in adm_counter.most_common(15)]

    for med in top15_adm_med:
        col_name = f"admmed_{_slugify(med)}"
        baseline_input[col_name] = adm_lists_norm.apply(
            lambda meds: int(med in set(meds))
        ).astype(int)

    baseline_input = baseline_input.drop(columns=['adm_med_name'])

    print('Top 15 adm_med_name medications (normalized):')
    print(top15_adm_med)
```
